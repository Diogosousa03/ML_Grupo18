### 1. Data Ingestion & Initial Inspection

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns 

from scipy.stats import chi2_contingency
```

```python
# Cramér's V function, used in the section 1.7.
def cramers_v(x, y):
    """Cramér's V (with bias correction) between two categorical variables: 0 means no association, 1 a perfect one."""
    cm = pd.crosstab(x, y)
    chi2 = chi2_contingency(cm)[0]
    n = cm.sum().sum()
    phi2 = chi2 / n
    r, k = cm.shape
    phi2corr = max(0, phi2 - ((k-1)*(r-1))/(n-1))
    rcorr = r - ((r-1)**2)/(n-1)
    kcorr = k - ((k-1)**2)/(n-1)
    if min((kcorr-1), (rcorr-1)) <= 0:
        return 0.0
    return np.sqrt(phi2corr / min((kcorr-1), (rcorr-1)))
```

### 1.6 Univariate Analysis 

*Univariate analysis studies one variable at a time to understand its characteristics and distribution*

```python
dist = pd.DataFrame({
    "n_customers": df["customer_since_year"].value_counts().sort_index(),
    "pct": (df["customer_since_year"].value_counts(normalize=True).sort_index() * 100).round(1)
})
display(dist)
dist["n_customers"].plot(
    kind="bar",
    figsize=(8, 4),
    title="Customers by customer_since_year",
    xlabel="Year",
    ylabel="Number of customers"
)
```

```python
# Skewness calculation (Asymmetric)
skew_value = df['customer_since_year'].skew()
print(f"Skewness of customer_since_year: {skew_value:.2f}")

if skew_value > 0:
    print("Conclusion: Asymmetrical distribution to the right (right-skewed).")
elif skew_value < 0:
    print("Conclusion: Asymmetrical distribution to the left (left-skewed).")
else:
    print("Conclusion: Approximately symmetric distribution.")
```

The mean (2017) is higher than the median (2015), so the distribution is right-skewed (skewness of 1.36), as the chart shows. This is not caused by outliers: 664 customers (58%) have `customer_since_year` = 2015, the minimum value, while the rest are spread over 2016–2026. Since 2015 is probably the year the records start, this value most likely groups all customers who joined in 2015 or earlier, which creates the peak on the left of the distribution.<br>
**Actionable:** the year is not log-transformed (on a year, log1p only maps 2015–2026 to 7.609–7.614), and the cleaning notebook adds the binary flag `since_2015_or_earlier` (2.4), so the model can tell these customers apart.

 ### 1.7 Bivariate Analysis  

```python
# Hierarchy check: does each municipality belong to only one commercial region?
check_nesting = df.groupby("municipality")["commercial_region"].nunique()
print("Strictly nested (each municipality in exactly 1 commercial region):", check_nesting.max() == 1)
```

```python
geo_cols = ['country_or_region', 'commercial_region', 'island', 'municipality']

matrix_v = pd.DataFrame(np.nan, index=geo_cols, columns=geo_cols)

for c in geo_cols:
    matrix_v.loc[c, c] = 1.0  # diagonal

# Lower triangle only: each pair is computed once (Cramér's V is symmetric)
for i, c1 in enumerate(geo_cols):
    for c2 in geo_cols[i + 1:]:
        pair = df[[c1, c2]].dropna()
        matrix_v.loc[c2, c1] = cramers_v(pair[c1], pair[c2])

plt.figure(figsize=(8, 6))
sns.heatmap(matrix_v, annot=True, fmt='.2f',
            cmap='Blues', vmin=0, vmax=1, square=True,
            cbar_kws={'pad': 0.1})
plt.title("Categorical Association Heatmap (Cramér's V)")
plt.yticks(rotation=0)
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

### 1.7.1 Geographic Hierarchy and Redundancy

Each cell of the heatmap is the Cramér's V between two geographic variables, from 0 (no association, light) to 1 (one variable determines the other, dark blue); only the lower triangle is filled because the measure is symmetric.

The Cramér's V heatmap shows very strong associations among `commercial_region`, `island` and `municipality` (0.91 to 0.99). This is expected: the geographic variables are strictly nested, and the nesting check above confirms that each municipality belongs to exactly one commercial region. However, Cramér's V tends towards 1 when a variable has many categories (183 municipalities for about 1000 customers), so these values indicate redundancy but overstate its exact strength.

`country_or_region` is almost perfectly associated with `commercial_region` (0.98), but shows 0.00 with `island` and `municipality`. This is not a real absence of association: `island` and `municipality` are only filled for CT10 customers, so after removing the missing values only one country is left, and a variable with a single category always gives 0.

**Implication:** these features are largely redundant and would introduce severe categorical multicollinearity if they were all used together. The choice is not made here: all the levels stay in the dataset as candidates, and the feature selection notebook decides which levels the model keeps.

```python
plt.figure(figsize=(10, 5))

# Select the top commercial regions for readability
top_reg = df['commercial_region'].value_counts().head(6).index
df_top_reg = df[df['commercial_region'].isin(top_reg)]

sns.boxplot(
    data=df_top_reg,
    x='commercial_region',
    y='customer_since_year',
    legend=False
)

plt.title("Customer Acquisition Year Distribution by Commercial Region (Top 6)")
plt.xlabel("Commercial Region")
plt.ylabel("Customer Since Year")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

### 1.7.2 Temporal Distribution across Commercial Regions

Each box covers the middle 50% of the years of entry of a region (from the lower to the upper quartile), the line inside is the median, the whiskers reach the furthest years within 1.5 × IQR and the dots are outliers.

Across the top six commercial regions, the median acquisition year is similar, between 2015 and 2017, so most regions share the same historical baseline. The differences are in the spread:

- **Concentrated, older base (RG17, RG18, RG27, RG31):** the median and the lower quartile are 2015 and the upper quartile is between 2016 and 2018, so most of their customers joined in 2015–2018, and the most recent ones appear as outliers.
- **Wider spread, more recent acquisition (RG23, RG29):** the median is 2017, the upper quartile reaches about 2020–2021 and the whiskers extend to 2026, with no outliers, so these regions kept acquiring customers for longer.

Overall, regions differ less in *when* they started than in *how long* they kept growing.

**Actionable:** no transformation is needed. `customer_since_year` and `commercial_region` stay as separate candidates, and the flag `since_2015_or_earlier` (cleaning notebook, 2.4) already marks the old base that dominates RG17, RG18, RG27 and RG31.
